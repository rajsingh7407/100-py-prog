## 1 · Why this matters

Every program has to **communicate**. It needs to show you what it found, calculated, or decided. The simplest way is `print()`, and you will use it every day for as long as you write Python. It's also your best debugging tool: when you don't know what your code is doing, print it.

Python runs your program **from top to bottom, one statement at a time**. Each line is carried out completely before the next one starts. That sounds obvious, but holding it in your head is the foundation of everything in this course.

## 2 · `print()`

`print()` displays whatever you put between its parentheses. Text goes inside quotes:

````python
print("Hello, world!")
````

Several `print()` calls run one after another, each starting on a new line:

````python
print("First line")
print("Second line")
print("Third line")
````

A `print()` with nothing inside prints an empty line, handy for spacing out your output:

````python
print("Above the gap")
print()
print("Below the gap")
````

### Text versus numbers

Anything inside quotes is **text** (programmers call it a *string*). Python prints it exactly as written and never tries to understand it. Numbers have **no quotes**, and Python knows how to do arithmetic with them.

Compare these:

````python
print("3 + 4")
print(3 + 4)
````

The first prints the characters `3 + 4`. The second *calculates* and prints `7`. The quotes make all the difference.

### Single or double quotes?

Either works, as long as the opening and closing quotes match:

````python
print("Double quotes")
print('Single quotes')
````

This lets you include one kind of quote inside the other:

````python
print("It's easy to include an apostrophe")
print('She said "hello" and waved')
````

## 3 · Printing several things at once

You can give `print()` several values separated by commas. It prints them on one line with a space between each:

````python
print("My age is", 30)
print("Total:", 4 + 5, "items")
````

Notice that you didn't have to write the space yourself, and that text and numbers mix freely.

### Controlling the separator with `sep`

The space between values is the **separator**. You can replace it using the `sep` option:

````python
print(2026, 10, 18, sep="-")
print("a", "b", "c", sep="")
print("one", "two", "three", sep=" | ")
````

### Controlling the line ending with `end`

Normally `print()` finishes with a **newline** (jump to the next line). The `end` option changes that:

````python
print("Loading", end="...")
print("done!")
````

````python
print("A", end="")
print("B", end="")
print("C")
````

`sep` goes **between** the values inside one `print()`. `end` goes **after** the last value. Both are written with a name and an equals sign, after the values.

## 4 · Comments

A **comment** is a note for human readers. Python ignores everything from a `#` to the end of the line.

````python
# This whole line is a comment: Python skips it.
print("Hello")  # A comment can also follow code on the same line.
````

### Why bother?

Code is read far more often than it is written, including by *you* in six months, when you've forgotten everything. Good comments explain **why**, not just what.

| Not helpful | Helpful |
|---|---|
| `# print the number` | `# Show the total before tax so the customer can compare` |
| `# add 1` | `# The first row is a header, so skip it` |

If a comment only repeats the code, delete it. If the code is confusing, first try to make the code clearer; then comment what remains.

### Commenting out code

Putting `#` in front of a line **disables** it without deleting it. Programmers do this all the time to test things:

````python
print("This line runs")
# print("This line does not")
print("This line runs too")
````

In a notebook you can comment or uncomment the selected lines with **Ctrl + /** (or **Cmd + /** on a Mac).

## 5 · Reading error messages

Sooner or later you'll type something Python can't understand. It then stops and shows an **error message**. Beginners often find these frightening, but they are the most helpful thing Python does for you: they tell you what went wrong and where.

Let's break something on purpose.

⚠️ **Expected to fail.** Run the next cell.

````python
print("Hello, world!)
````

The red box is called a **traceback**. Read it like this:

```
  Cell In[1], line 1            ← where: which cell and line
    print("Hello, world!)       ← the offending line
          ^                     ← an arrow pointing at roughly where Python got lost
SyntaxError: unterminated string literal (detected at line 1)    ← what: type of error + explanation
```

**Read the last line first.** It names the *kind* of error (`SyntaxError`) and explains it in plain words: we opened a string with `"` but never closed it.

There are two kinds of error you'll meet constantly in this notebook.

### `SyntaxError`: Python can't make sense of the code

A syntax error means the code is not written according to Python's grammar, like a sentence with missing punctuation. Python can tell **before running anything**. Common causes: a missing quote or bracket, or a stray character.

⚠️ **Expected to fail.** A missing closing bracket:

````python
print("Hello"
````

Notice the explanation: `'(' was never closed`. Python tells you exactly which bracket is the problem.

### `NameError`: Python doesn't know that word

If you write a word without quotes, Python assumes it's a **name** it should already know, such as `print`. If it isn't, you get a `NameError`.

⚠️ **Expected to fail.** Forgetting the quotes around text:

````python
print(Hello)
````

Python reads `Hello` as the name of something. It has never heard of anything called `Hello`, hence `NameError: name 'Hello' is not defined`.

> 💡 **Rule of thumb.** Text needs quotes. Names and numbers don't.

### A recipe for reading any error

1. Read the **last line**: error type and message.
2. Look at the **line number** and the arrow.
3. Ask: what did I *mean* to write there? Compare with what's actually there.
4. Fix one thing, run again. Don't change five things at once.

Python's error messages have become friendlier with every release, and often include suggestions such as *"Did you mean: 'print'?"*.

## 6 · Python as a calculator (a preview)

Python does arithmetic. Here are the basic operators; we'll study them properly in notebook 03.

| Symbol | Meaning |
|---|---|
| `+` | add |
| `-` | subtract |
| `*` | multiply |
| `/` | divide |
| `**` | to the power of |
| `( )` | group, as in school maths |

````python
print(2 + 3)
print(10 - 4)
print(6 * 7)
print(15 / 4)
print(2 ** 10)
print((2 + 3) * 4)
````

Python follows the usual order of operations, so `*` happens before `+`. Use brackets to be explicit.

### A notebook shortcut

In a notebook, if the **last line** of a cell is an expression, the notebook shows its value even without `print()`:

````python
2 + 3
````

But only the last line is shown, and **this is a notebook feature, not Python itself**. In a regular `.py` script nothing would appear. Get in the habit of using `print()` for output you care about.

````python
10 * 10
20 * 20
````

Only `400` appears, because `100` was not the last line. Use `print()` for each value you want to see.

## 7 · Common mistakes

Run each cell and read the error. ⚠️ **Expected to fail**, apart from the last fix.

**1. Forgetting the parentheses.** Old versions of Python (and many websites) used `print "hello"`. Python 3 needs brackets.

````python
print "hello"
````

**2. Capital letters matter.** Python is *case-sensitive*: `print` and `Print` are different names.

````python
Print("hello")
````

**3. Mixing quote types.** The opening and closing quote must match.

````python
print('hello")
````

**4. A stray space at the start of a line.** Indentation means something in Python (we'll use it in notebook 08), so you can't just start a line with a space.

````python
 print("hello")
````

**5. Curly "smart" quotes.** If you copy code from a web page or a word processor, the quotes may be curly. Python only accepts straight ones.

````python
print(“hello”)
````

**6. Adding text and numbers.** You can't use `+` to glue text onto a number. Use a comma in `print()` instead.

````python
print("Total: " + 5)
````

````python
print("Total:", 5)
````
