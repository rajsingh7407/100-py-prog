## 1 · Why this matters

A **program** is a list of precise instructions that a computer follows. A recipe is a good analogy: "boil 2 cups of water, add the pasta, wait 10 minutes". A person can fill in gaps ("until it is soft enough"); a computer cannot. It does *exactly* what you write, in the order you write it, and it does it millions of times faster than you could.

That combination (**exact** and **fast**) is what makes programming powerful. Anything you can describe precisely can be automated: renaming a thousand files, analysing a year of sales, training a neural network, running a website.

### What is Python?

**Python** is a programming language created by Guido van Rossum, who started it as a hobby project in December 1989 and released it in 1991. The name comes from the British comedy group *Monty Python's Flying Circus*, not the snake, which is why you will see jokes about spam and parrots in Python documentation.

Python was designed to be **readable**: code often looks close to plain English. That makes it a great first language, and it is also one of the most widely used languages in the world today:

| Area | What Python is used for |
|---|---|
| Data and science | analysing data, plotting, scientific computing |
| AI and machine learning | neural networks, language models (the companion courses use Python) |
| Web | websites and web services |
| Automation | scripts that do boring jobs for you |
| Education | the first language at many universities |

The current major version is **Python 3** (released in 2008). This course uses **Python 3.12**.

## 4 · The kernel, and why run order matters

Behind every notebook is a **kernel**: a running Python process that remembers everything you have done in it. The notebook is just the page you type into; the kernel is the thing that actually runs the code.

Because the kernel has a *memory*, what a cell does can depend on which cells you ran before it. This is the number one source of confusion for beginners, so let's see it happen.

(The line `lives = 3` below stores the number 3 under the name `lives`. You'll learn all about this in notebook 02. For now just watch what happens.)

````python
lives = 3
````

````python
lives = lives - 1
print("Lives left:", lives)
````

**Try it:** run the second cell above again, and again. Each time it prints a smaller number, because the kernel *remembers* the value of `lives` between runs. Now go back and run `lives = 3` again, then the second cell once. It's back to 2.

The code in the cell didn't change, but the result did. What matters is **the order you run cells in, not the order they appear on the page**.

### Restarting the kernel

If your notebook gets into a muddled state (you ran cells out of order, deleted a cell but its variable is still alive, something seems wrong for no reason), the fix is to restart:

- Menu **Kernel → Restart Kernel and Run All Cells…** wipes the kernel's memory and runs the notebook from top to bottom.
- Menu **Kernel → Restart Kernel…** wipes the memory and leaves the cells alone.
- Menu **Kernel → Interrupt Kernel** (or the ■ button) stops code that is running for too long, such as a loop that never ends.

When something strange happens, "restart and run all" is the first thing to try.

> ⚠️ **Heads-up about errors.** Some cells in this course contain code that is *meant* to fail, so that you learn to read error messages. They are marked with **"⚠️ Expected to fail"**. When Jupyter's *Run All* reaches an error it stops, so for those notebooks run cells one at a time with Shift + Enter, which is how you should work anyway.

### Troubleshooting checklist

| Symptom | What to try |
|---|---|
| `uv: command not found` | Close and reopen the terminal. If it still fails, rerun the install command. |
| `uv run jupyter lab` says it cannot find a project | You are in the wrong folder. `cd` into `python-course` (the folder containing `pyproject.toml`). |
| The browser does not open | Copy the `http://localhost:8888/...` link from the terminal output into your browser. |
| Notebook says "No kernel" or asks you to pick one | Choose **Python 3 (ipykernel)** in the kernel picker (top right of the notebook). |
| `ModuleNotFoundError: No module named 'course_tools'` | Run the setup cell at the top of the notebook (see below), and make sure you opened the notebook from the `notebooks` folder. |
| Code seems to ignore your changes | You edited but did not re-run the cell. Press Shift + Enter. Still odd? Restart the kernel. |
| A cell shows `[*]` forever | It is running (or waiting for input). Use **Kernel → Interrupt Kernel**. |
| Port already in use | Another Jupyter is running. Close it, or use `uv run jupyter lab --port 8889`. |

## 7 · Common mistakes

Everybody does these in their first week:

1. **Forgetting to run the cell.** Editing the code does nothing until you press Shift + Enter.
2. **Running cells out of order.** Cell 5 uses something cell 3 creates, but you skipped cell 3. Fix: restart and run all.
3. **Copy-pasting "smart quotes".** Python needs straight quotes `"` and `'`, not the curly ones `“ ”` that word processors produce.
4. **Typing code into a Markdown cell** (it just shows as text), or text into a code cell (it produces an error). The cell-type dropdown in the toolbar shows which kind a cell is.
5. **Being scared of red error text.** Errors are Python telling you what's wrong. Reading them is a skill, and notebook 01 teaches it.
