Python is an excellent calculator, but it has a few quirks that bite beginners and professionals alike. Knowing about them now will save you hours later.

## 2 · Two kinds of numbers

Python has two main number types:

- **`int`**: whole numbers (integers): `7`, `-3`, `0`, `1_000_000`
- **`float`**: numbers with a decimal point ("floating-point"): `3.14`, `-0.5`, `2.0`, `1e6`

`2` and `2.0` look equal, but they are different types. The decimal point is what makes a float.

````python
print(type(7))
print(type(7.0))
print(type(1e6))      # scientific notation: 1 x 10^6, always a float
print(1_000_000)      # underscores are allowed in numbers to make them readable
````

Python integers can be **as large as you like**. There's no overflow like in some other languages:

````python
print(2 ** 100)
print(10 ** 30 + 1)
````

## 3 · The arithmetic operators

| Operator | Name | Example | Result |
|---|---|---|---|
| `+` | addition | `7 + 2` | `9` |
| `-` | subtraction | `7 - 2` | `5` |
| `*` | multiplication | `7 * 2` | `14` |
| `/` | true division | `7 / 2` | `3.5` |
| `//` | floor division | `7 // 2` | `3` |
| `%` | modulo (remainder) | `7 % 2` | `1` |
| `**` | power | `7 ** 2` | `49` |

````python
print(7 + 2, 7 - 2, 7 * 2)
print(7 / 2, 7 // 2, 7 % 2)
print(7 ** 2)
````

### Division, three ways

This is where most of the surprises are, so let's slow down.

- `/` **always** gives a `float`, even when the answer is whole: `8 / 2` is `4.0`.
- `//` is **floor division**: divide, then round *down* to a whole number: `7 // 2` is `3`.
- `%` is the **remainder** after floor division: `7 % 2` is `1`, because 2 goes into 7 three times with 1 left over.

They fit together: `7 == (7 // 2) * 2 + (7 % 2)`, i.e. `7 == 3 * 2 + 1`.

````python
print(8 / 2)
print(17 // 5, 17 % 5)      # 17 = 3 * 5 + 2
print(-7 // 2)              # "down" means towards negative infinity, so -4, not -3
````

### What is `%` good for?

The remainder operator is surprisingly useful:

- **Even or odd:** `n % 2` is `0` for even numbers, `1` for odd.
- **Divisibility:** `n % 7 == 0` means `n` is a multiple of 7.
- **Wrapping around:** hours on a clock, days of the week, anything cyclic.
- **Extracting digits:** `1234 % 10` is `4` (the last digit), and `1234 // 10` is `123` (everything else).

````python
print(10 % 2, 11 % 2)
print(1234 % 10, 1234 // 10)

# What time is it 50 hours after 9 o'clock?  (on a 24-hour clock)
print((9 + 50) % 24)
````

### Power

`**` raises to a power. Because a square root is a power of `0.5`, you can use it for that too.

````python
print(2 ** 10)
print(9 ** 0.5)     # square root
print(10 ** -2)     # negative powers give fractions: 1 / 100
````

## 4 · Order of operations

Python follows the same precedence as school maths. From highest priority to lowest:

| Priority | Operators |
|---|---|
| 1 (first) | `( )` parentheses |
| 2 | `**` |
| 3 | unary minus, e.g. `-x` |
| 4 | `*  /  //  %` |
| 5 (last) | `+  -` |

Operators of the same priority run **left to right**, except `**`, which runs right to left.

````python
print(2 + 3 * 4)        # multiplication first
print((2 + 3) * 4)      # parentheses first
print(100 / 10 * 5)     # left to right: (100 / 10) * 5
print(2 ** 3 ** 2)      # right to left: 2 ** (3 ** 2) = 2 ** 9
````

One more famous trap: `**` binds tighter than the minus sign.

````python
print(-3 ** 2)          # this is -(3 ** 2)
print((-3) ** 2)
````

> 💡 **When in doubt, add parentheses.** They cost nothing and make your intent clear to the next reader (who might be you).

### Mixing ints and floats

If an operation mixes an `int` and a `float`, the result is a `float`. If both are `int`, the result is an `int`, except for `/`, which always gives a `float`.

````python
print(3 + 2.0)
print(3 + 2)
print(6 / 3)
print(6 // 3)
print(6.0 // 3)
````

The functions `int()` and `float()` convert between the types. `int()` **chops off** the decimal part (towards zero), it doesn't round.

````python
print(int(3.9))
print(int(-3.9))
print(float(5))
````

## 5 · Why `0.1 + 0.2` is not `0.3`

Run this and be surprised:

````python
print(0.1 + 0.2)
print(0.1 + 0.2 == 0.3)
````

Python isn't broken. This happens in every programming language, and the reason is worth understanding.

Computers store numbers in **binary** (base 2). Just as `1/3` can't be written exactly in decimal (`0.3333...` goes on forever), a number like `0.1` can't be written exactly in binary: it's `0.0001100110011...` forever. The computer has to **cut it off** after about 16-17 significant digits. So `0.1` is stored as a number *extremely close to* `0.1`, but not exactly `0.1`. Add two such approximations and the tiny errors show up.

You can peek at what's really stored by asking for lots of digits (don't worry about the format syntax; we study it in notebook 05):

````python
print(f"{0.1:.30f}")
print(f"{0.1 + 0.2:.30f}")
````

### What to do about it

1. **Don't compare floats with `==`.** Check that they are *close enough* instead, with `math.isclose()`.
2. **Round for display.** `round(x, 2)` keeps two decimal places, which hides the noise.
3. **For money, be careful.** Real systems keep amounts as whole numbers of cents (an `int`), or use Python's `decimal` module, which does exact decimal arithmetic. We'll stick to ints and rounding in this course.

````python
import math

print(math.isclose(0.1 + 0.2, 0.3))
print(round(0.1 + 0.2, 2) == 0.3)

# money as whole cents is exact
price_in_cents = 1999
print(price_in_cents * 3)
````

Note that integers are always exact. The problem is only with floats.

## 6 · Handy built-in functions

These work out of the box, with no import needed.

| Function | What it does | Example |
|---|---|---|
| `abs(x)` | distance from zero (absolute value) | `abs(-4.5)` → `4.5` |
| `round(x)` | nearest integer | `round(3.7)` → `4` |
| `round(x, n)` | round to `n` decimal places | `round(3.14159, 2)` → `3.14` |
| `min(a, b, ...)` | smallest of the values | `min(4, 9, 2)` → `2` |
| `max(a, b, ...)` | largest of the values | `max(4, 9, 2)` → `9` |
| `pow(a, b)` | same as `a ** b` | `pow(2, 10)` → `1024` |
| `divmod(a, b)` | `(a // b, a % b)` together | `divmod(17, 5)` → `(3, 2)` |

````python
print(abs(-4.5))
print(round(3.7), round(3.14159, 2))
print(min(4, 9, 2), max(4, 9, 2))
print(divmod(17, 5))
````

### A rounding surprise

Python rounds halves to the **nearest even** number (called "banker's rounding", it avoids a bias that would add up over many roundings):

````python
print(round(0.5), round(1.5), round(2.5), round(3.5))
````

So `round(2.5)` is `2`, not `3`. If you need "school" rounding (halves always up), you have to do something extra. It rarely matters, but it's good to know it's deliberate. Also, `round()` of a float is subject to float noise: `round(2.675, 2)` gives `2.67`, because `2.675` is stored as slightly less than `2.675`.

## 7 · The `math` module

Python comes with a large **standard library** of extra tools, organised into **modules**. To use one, `import` it. Then access its contents with a dot.

````python
import math

print(math.sqrt(144))
print(math.pi)
print(math.floor(2.7), math.ceil(2.1))   # round down / round up
print(math.hypot(3, 4))                  # length of the hypotenuse
print(math.isclose(0.1 + 0.2, 0.3))
````

`math` has dozens more: `sin`, `cos`, `log`, `factorial`, `gcd`... To see what's available, ask for help. You can do this for any module or function; reading docs is a skill worth practising early.

````python
help(math.floor)
````

## 8 · Augmented assignment

Updating a variable based on its current value is so common that Python has shortcuts:

| Long form | Short form |
|---|---|
| `x = x + 5` | `x += 5` |
| `x = x - 5` | `x -= 5` |
| `x = x * 5` | `x *= 5` |
| `x = x / 5` | `x /= 5` |
| `x = x // 5` | `x //= 5` |
| `x = x % 5` | `x %= 5` |
| `x = x ** 5` | `x **= 5` |

````python
score = 10
score += 5       # 15
score *= 2       # 30
score -= 1       # 29
print(score)
````

There is no `x++` in Python; use `x += 1`.

## 9 · Common mistakes

### 1. Division by zero

⚠️ **Expected to fail.**

````python
print(10 / 0)
````

`ZeroDivisionError` is Python's way of saying "that is undefined". `//` and `%` fail the same way.

### 2. `^` is not power

In some languages and on calculators `^` means "to the power of". In Python it's something completely different (a bit-twiddling operator), so you get a plausible but wrong answer with no error!

````python
print(2 ^ 3)     # not 8!
print(2 ** 3)    # this is the power operator
````

### 3. Implied multiplication

In maths, `2(3 + 4)` means `2 × 7`. In Python you must write the `*` yourself. ⚠️ **Expected to fail.**

````python
print(2(3 + 4))
````

### 4. Forgetting that `/` gives a float

If you want a whole number from a division, use `//` (or convert with `int()`).

````python
people = 12
groups = people / 4
print(groups)            # 3.0, not 3
print(people // 4)       # 3
````

### 5. Comparing floats with `==`

You saw it above: `0.1 + 0.2 == 0.3` is `False`. Use `math.isclose()` or round first.

### 6. Leading zeros

Python 3 doesn't allow numbers like `007`; there's no meaning for a leading zero. ⚠️ **Expected to fail.**

````python
print(007)
````
