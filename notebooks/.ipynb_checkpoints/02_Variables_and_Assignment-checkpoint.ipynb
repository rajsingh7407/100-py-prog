## 1 · Why this matters

So far every value in our programs has been typed directly into the code. That gets tedious and fragile very quickly. Suppose you are computing a price:

```python
print(19.99 * 3)
print(19.99 * 3 * 0.2)
print(19.99 * 3 * 1.2)
```

The number `19.99` appears three times. If the price changes, you must find and fix every copy, and miss one and your program quietly gives wrong answers. And what does `0.2` mean? A reader has to guess.

A **variable** solves both problems: it gives a value a **name**. Write the value once, use the name everywhere, and the name can say what the value *means*.

````python
price = 19.99
quantity = 3
tax_rate = 0.2

subtotal = price * quantity
tax = subtotal * tax_rate
total = subtotal + tax

print("Subtotal:", subtotal)
print("Tax:", tax)
print("Total:", total)
````

Change `price` to `24.50` and re-run the cell: everything updates. This is *much* easier to read, too.

(You may notice a long tail of digits on some of the results. That's normal for decimal numbers on computers, and notebook 03 explains it.)

## 2 · Assignment

The line `price = 19.99` is an **assignment statement**. It reads:

> *Take the value on the right and store it under the name on the left.*

It is **not** a maths equation! `=` here means "store", not "is equal to". The value goes **right to left**.

Python carries out an assignment in two steps:

1. Work out the value on the **right-hand side**.
2. Store that value under the name on the **left-hand side**.

A variable doesn't print anything when you assign to it; you use it later by writing its name:

````python
planet = "Mars"
moons = 2

print(planet)
print(planet, "has", moons, "moons")
````

### The "labelled box" mental model

Picture each variable as a **labelled box** that holds one value:

```
   ┌──────────┐      ┌──────────┐
   │  "Mars"  │      │    2     │
   └──────────┘      └──────────┘
      planet            moons
```

Assigning puts a value into the box with that label. Using the name looks inside the box. (Later in the course we will refine this picture slightly: it is really more like a *name tag* attached to a value. For now the box works fine.)

### Reassignment

You can put a new value in the box at any time. The old value is thrown away:

````python
score = 10
print(score)

score = 25
print(score)
````

### Using a variable to update itself

Because the right side is worked out *first*, a variable can appear on both sides:

````python
score = 10
score = score + 5
print(score)
````

Read `score = score + 5` as: "take the current value of `score`, add 5, and store the result back in `score`." In algebra `x = x + 5` would be nonsense, but in programming it's an everyday move.

### Tracing a program

When a program confuses you, **trace it**: go line by line, writing down each variable's value after each line. Here's a program and its trace:

````python
a = 4
b = 7
a = a + b
b = a - b
print(a, b)
````

| After line | `a` | `b` |
|---|---|---|
| `a = 4` | 4 | – |
| `b = 7` | 4 | 7 |
| `a = a + b` | **11** | 7 |
| `b = a - b` | 11 | **4** |

So it prints `11 4`. Notice that the values only change on the line that assigns to them. A pencil and a table like this will solve many mysteries.

## 3 · Naming variables

Python has a few **rules** (break them and you get a `SyntaxError`), and the community has some **conventions** (break them and your code is just harder to read).

### The rules

- Names can contain **letters, digits and underscores** (`_`).
- They **cannot start with a digit**: `2nd_place` is invalid, `place_2` is fine.
- **No spaces or hyphens**: `total cost` and `total-cost` are invalid.
- Names are **case-sensitive**: `Score`, `score` and `SCORE` are three different variables.
- They can't be one of Python's **reserved words**, the words that already mean something in the language. Here's the full list (the first line borrows a tool from Python's standard library with `import`; notebook 03 explains how):

````python
import keyword

print(keyword.kwlist)
````

You'll meet most of these during Level 1 (`if`, `else`, `while`, `for`, `True`, `None`, ...).

### The conventions

- Use **`snake_case`**: lowercase words joined by underscores: `first_name`, `total_price`.
- Choose **descriptive** names. `seconds_per_day` beats `spd`, which beats `x`. Short names such as `i` or `n` are fine for tiny, obvious uses.
- Names for values that never change are written in `UPPER_CASE`: `SPEED_OF_LIGHT = 299_792_458`. Python doesn't enforce it, but readers will understand your intent.
- Don't reuse the names of Python's built-in functions (like `print`, `type`, `max`, `sum`, `input`). It's legal, but it hides the real thing. We'll see the damage in the Common mistakes section.

````python
# Good
first_name = "Ada"
days_in_week = 7
SPEED_OF_LIGHT = 299_792_458

# Legal but unhelpful
x1 = "Ada"
d = 7
````

## 4 · Values have types

Every value in Python has a **type**, which tells Python what kind of thing it is and what you can do with it. `type()` reveals it:

````python
print(type(42))
print(type(3.14))
print(type("hello"))
print(type(True))
````

These four are the types you'll use most:

| Type | Meaning | Examples |
|---|---|---|
| `int` | whole numbers | `42`, `-7`, `0` |
| `float` | numbers with a decimal point | `3.14`, `-0.5`, `2.0` |
| `str` | text ("string") | `"hello"`, `'x'`, `""` |
| `bool` | true or false | `True`, `False` |

Note that `42` (an `int`), `42.0` (a `float`) and `"42"` (a `str`) are **three different values** that happen to look alike.

````python
print(type(42))
print(type(42.0))
print(type("42"))
````

It's the **value** that has a type; the variable is just a label, and you can attach a different kind of value to it later:

````python
thing = 5
print(type(thing))

thing = "five"
print(type(thing))
````

This flexibility is called *dynamic typing*. Handy, but don't overdo it: a variable that is a number on one line and text on the next makes code hard to follow. Usually a variable should hold the same *kind* of thing for its whole life.

There's also a special "nothing" value, `None`, with its own type `NoneType`. It stands for "no value here yet". More on that in notebook 07.

````python
nothing_yet = None
print(nothing_yet)
print(type(nothing_yet))
````

## 5 · Assigning several variables at once

You can assign several variables in one line:

````python
width, height = 4, 7
print(width, height)
````

The values on the right are matched to the names on the left in order. You can also give several names the same value:

````python
x = y = z = 0
print(x, y, z)
````

### Swapping two values

A very common task: exchange the contents of two variables. This does **not** work:

````python
a = "left"
b = "right"

a = b
b = a
print(a, b)
````

Trace it: after `a = b`, both boxes contain `"right"`: the old `"left"` is gone, so `b = a` copies the new `"right"` back.

**Fix 1: a temporary variable**, like pouring a third glass of water to swap two full glasses:

````python
a = "left"
b = "right"

temp = a
a = b
b = temp
print(a, b)
````

**Fix 2: Python's multiple assignment**, which works out the entire right side *before* assigning, so nothing gets lost:

````python
a = "left"
b = "right"

a, b = b, a
print(a, b)
````

That one-liner is how Python programmers swap, and you'll see it often.

## 6 · Common mistakes

### 1. Using a name before it has a value

Python can only use names it has already seen assigned. ⚠️ **Expected to fail.**

````python
print(favourite_colour)
````

`NameError` again. Check the spelling, check that the assignment ran, and check that you ran the cell that contains it (see notebook 00!). A typo creates the same error. ⚠️ **Expected to fail.**

````python
number_of_apples = 5
print(number_of_appels)
````

### 2. `=` versus `==`

`=` **stores** a value. `==` **asks a question**: "are these two values equal?" The answer is a `bool`, `True` or `False`.

````python
x = 5          # store 5 in x
print(x == 5)  # ask: is x equal to 5?
print(x == 6)
````

Mixing them up is one of the most common beginner errors. ⚠️ **Expected to fail.** You can't assign to a number:

````python
5 = x
````

### 3. A variable does not remember how it was calculated

This one catches nearly everyone. Predict what this prints before running it:

````python
price = 10
quantity = 3
total = price * quantity

price = 20          # the price changes...
print(total)        # ...but total was calculated once, already
````

`total` stays `30`. The line `total = price * quantity` calculated a number **at that moment** and stored it. Python doesn't keep an "equation" linking `total` to `price`. (A spreadsheet does, but a program doesn't.) If you want the new total, you have to run the calculation again:

````python
total = price * quantity
print(total)
````

### 4. Naming a variable after a built-in

You can do it, but then you can't use the original. ⚠️ **Expected to fail.**

````python
type = "box"
print(type(5))
````

We replaced Python's `type` with a string, and a string can't be called like a function. To get the original back, delete our version of the name with `del`:

````python
del type
print(type(5))
````

### 5. Case and spelling

`Score` is not `score`. Python won't correct you. If you get a `NameError` for something you are sure you defined, compare the spelling character by character.

## 8 · Recap and what's next

- A **variable** is a name for a value. `name = value` stores the value, working out the right side first.
- Picture a **labelled box**; reassigning replaces the contents. `x = x + 1` is perfectly normal.
- **Trace** programs line by line with a table when you're unsure.
- Names use letters, digits and underscores, can't start with a digit, can't be reserved words, and are case-sensitive. Use descriptive `snake_case`.
- Every value has a **type**: `int`, `float`, `str`, `bool`, ... (`type()` shows it).
- `a, b = b, a` swaps two values; `a, b = 1, 2` assigns several at once.
- Traps: using a name before assigning it, `=` vs `==`, and expecting a variable to update by itself when the things it was calculated from change.
